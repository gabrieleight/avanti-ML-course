#### 01 - Explique, com suas palavras, o que é machine learning?

Machine Learning, ou aprendizado de máquina, é uma área da visão computacional que compreende o uso de redes neurais para identificar padrões, tomar decisões e fazer previsões a partir de um grande conjunto de dados

#### 02 - Explique o conceito de conjunto de treinamento, conjunto de validação e conjunto de teste em machine learning.

Conjunto de treinamento é o conjunto de dados usado para fazer treinamento do modelo.

Conjunto de validação é um conjunto de dados separado do conjunto de treinamento, usado para ajustar hiperparâmetros e verificar se o modelo está generalizando bem as previsões e evitar erros de underfitting ou overfitting.

Conjunto de teste é o conjunto de dados usado no teste de desempenho, depois que o modelo foi treinado e ajustado, para verificar o desempenho a partir de dados novos.

#### 03 - Explique como você lidaria com dados ausentes em um conjunto de dados de treinamento.

Algumas estratégias interessantes são, a partir do conjunto de dados, realizar algumas manipulações, como remover linhas e colunas ausentes usando métodos da biblioteca pandas, ou realizar o preenchimento de valores a partir de valores estatísticos ou usando modelos preditivos. Vale lembrar a origem dessa ausência de dados, e que deve-se descobrir o porque deles serem ausentes.

#### 04 - O que é uma matriz de confusão e como ela é usada para avaliar o desempenho de um modelo preditivo?

Uma matriz de confusão é uma tabela de comparação entre as previsões feitas pelo modelo com os valores verdadeiros ou reais, de maneira que mostra o quão eficiente o modelo foi em acertar ou errar as previsões. Em cima dessa matriz, é possível extrair algumas métricas de desempenho como a acurácia e precisão do modelo, F1-score e revocação.

#### 05 - Em quais áreas (tais como construção civil, agricultura, saúde, manufatura, entre outras) você acha mais interessante aplicar algoritmos de machine learning?

Uma das áreas mais interessantes e promissoras, sem dúvidas, é a área da saúde.
